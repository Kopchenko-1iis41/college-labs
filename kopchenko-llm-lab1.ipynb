Задание 10

```python
train['FamilySize'] = train['SibSp'] + train['Parch'] + 1
```

```python
train['IsAlone'] = train['FamilySize'].apply(lambda x: 1 if x == 1 else 0)
```

```python
train['AgeGroup'] = pd.cut(train['Age'], bins=[0, 12, 18, 35, 60, 100], labels=['ребёнок', 'подросток', 'молодой', 'взрослый', 'пожилой'])
```

```python
train
```

```python
train_age=train.groupby('AgeGroup')['Fare'].agg(['mean','count'])
train_age['Survived']=train.groupby('AgeGroup')['Survived'].mean()
train_age
```

```python
train.pivot_table(values='Survived',
 index='AgeGroup',
 columns='Sex',
aggfunc='mean')
```

Больше всего выжило детей от 0 до 12 лет, а также женщин всех возростных груп

Задание 11

```python
train_pt=train.groupby('Embarked')['PassengerId'].count().sort_values(ascending=False)
train_pt
```

```python
train_pt=train.groupby('Embarked')['Age'].agg(['count','mean'])
train_pt['Fare']=train.groupby('Embarked')['Fare'].mean()
train_pt['Survived']=train.groupby('Embarked')['Survived'].mean()
train_c1=train[train.Pclass==1]
train_pt['1class']=train_c1.groupby('Embarked')['Pclass'].count()
train_pt['Mean1Class']=train_pt['1class']/train_pt['count']*100
train_pt.sort_values(by='count',ascending=False)
```

Больше всего выживших в порту C потому, что половина пассажиров оттуда в 1-ом классе
